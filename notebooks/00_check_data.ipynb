```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

DATA_ROOT = os.path.join(os.pardir, "data", "raw")
df = pd.read_csv(os.path.join(DATA_ROOT, "train_labels.csv"))
df.head(1000)
```

```python
attack_samples = df[df["label"] == 1].sample(10)

fig, axes = plt.subplots(2, 5, figsize=(20, 8))
for ax, (_, row) in zip(axes.flat, attack_samples.iterrows()):
    img_path = os.path.join(DATA_ROOT, "train", "train", f"{row['id']}.jpeg")
    img = Image.open(img_path)
    ax.imshow(img)
    ax.set_title(f"{row['type']}\n{row['id'][:12]}...", fontsize=9)
    ax.axis("off")

fig.suptitle("10 Attack Images (label=1)", fontsize=14, fontweight="bold")
plt.tight_layout()
plt.show()
```
